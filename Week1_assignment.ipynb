```python
#printing clockwise elements
def print_clockwise(matrix):
    if not matrix.size:
        return
    
    top, bottom = 0, matrix.shape[0] - 1
    left, right = 0, matrix.shape[1] - 1
    
    while top <= bottom and left <= right:
        for i in range(left, right + 1):
            print(matrix[top][i], end=' ')
        top += 1
        
        for i in range(top, bottom + 1):
            print(matrix[i][right], end=' ')
        right -= 1
        
        if top <= bottom:
            for i in range(right, left - 1, -1):
                print(matrix[bottom][i], end=' ')
            bottom -= 1
            
        if left <= right:
            for i in range(bottom, top - 1, -1):
                print(matrix[i][left], end=' ')
            left += 1
```

```python
def numpy_alternate_sort(array):
    sorted_array = np.sort(array)
    n = len(sorted_array)
    
    result = np.empty(n, dtype=sorted_array.dtype)
    result[0::2] = sorted_array[:(n + 1) // 2]  # Even indices
    result[1::2] = sorted_array[(n + 1) // 2:][::-1]  # Odd indices
    
    return result
print(np.round(numpy_alternate_sort(array),2))
```

```python
#Creating the data set of scores of 10 students in 5 subjects
Names = ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack']
Subjects = ['Math', 'Science', 'English']
scores = np.random.randint(50, 100, 10)
import pandas as pd
data ={
    'Names': Names,
    #selecting random subjects for each student
    'Subjects': random.choices(Subjects, k=10),
    'Scores': scores

}
df = pd.DataFrame(data)
print(df)
```

```python
#adding a new column for the grade based on scores
def assign_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'
df['Grade'] = df['Scores'].apply(assign_grade)
print(df)
```

```python
new_data=df.sort_values(by='Scores', ascending=False, inplace=False)
print(new_data)
```

```python
#avg score of each subject
avg_scores = df.groupby('Subjects')['Scores'].mean().reset_index()
print(avg_scores)
```

```python
def pandas_filter(dataframe, threshold):
    return dataframe[dataframe['Scores'] > threshold]
def grade_to_threshold(grade):
    if grade == 'A':
        return 90
    elif grade == 'B':
        return 80
    elif grade == 'C':
        return 70
    elif grade == 'D':
        return 60
    else:
        return 0
print(pandas_filter(df, grade_to_threshold('B')))
```
